### 📘 Atividade Complementar – Estudo de Caso em PHP<br>
##### 🎯 Objetivo
#### Aplicar os conceitos de **formulários HTML, requisições via método POST e processamento em PHP** para desenvolver uma aplicação web simples.

##### 📖 Estudo de Caso
#### Uma clínica de nutrição deseja disponibilizar em seu site uma Calculadora de IMC (Índice de Massa Corporal), para que os pacientes possam verificar se estão dentro do peso adequado.

#####  **IMC** é calculado pela fórmula:

$$ IMC = \frac{peso}{altura^2} $$

##### IMC = altura2 peso
##### Classificação (OMS):

* **Menor que 18,5** -> Abaixo do peso
* **18,5 a 24,9** -> Peso normal
* **25,0 a 29,9** -> Sobrepeso
* **30,0 ou mais** -> Obesidade
