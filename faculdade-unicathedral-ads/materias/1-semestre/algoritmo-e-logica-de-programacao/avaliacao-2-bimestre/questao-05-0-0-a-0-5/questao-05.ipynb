### 05. Um professor deseja identificar quantos alunos obtiveram nota acima de 7 em uma avaliação. Para isso, ele criou um algoritmo que armazena as notas dos 6 alunos em um vetor e conta quantos estão acima da média esperada. No entanto, o trecho principal do pseudocódigo foi parcialmente apagado: Assinale a alternativa que completa corretamente o trecho faltante no algoritmo:
--------
### A)
### Se notas[i] > 7 então 
### ------ cont_aprovados <- cont_aprovados + 1
### FimSe

### B)
### Enquanto notas[i] > 7 faça
### ------ cont_aprovados <- cont_aprovados + 1
### FimEnquanto

### C)
### Se cont_aprovados > 7 então
### ------ notas[i] <- notas[i] + 1
### FimSe

### D)
### Para notas[i] de 1 até 6 faça
### ------ cont_aprovados <- cont_aprovados + 1
### FimPara

### E)
### Se notas[i] >= 7
### ------ então i-i+1
### FimSe

### **Resposta correta: (A)**
