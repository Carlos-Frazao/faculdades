### 05. Uma startup de tecnologia ambiental desenvolveu um sistema para monitorar os níveis de poluição em diferentes regiões de uma cidade. Os sensores instalados registram, diariamente, os valores de poluição do ar (em partículas por milhão ppm), que são armazenados em um **vetor de 10 posições**. Um trecho do algoritmo responsável por analisar os dados é o seguinte:

--------
### A) O vetor **poluicao** armazena os dados como texto e precisa ser convertido para número antes de comparações
### B) A estrutura condicional está inadequada, pois deveria ser usada dentro de um enquanto, e não de um para.
### C) O algoritmo calcula corretamente quantos dias apresentaram poluição acima de 50 ppm.
### D) A variável **dias_acima** armazena os valores de poluição acima de 50, e não a quantidade de dias
### E) O algoritmo possui erro, pois vetores não podem ser acessados com indices inteiros em pseudocódigo

### **Resposta correta: (C)**
