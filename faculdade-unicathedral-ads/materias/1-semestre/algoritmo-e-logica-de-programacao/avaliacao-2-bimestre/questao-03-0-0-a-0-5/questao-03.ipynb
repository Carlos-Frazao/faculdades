Para isso, eles registram a medição a cada hora, totalizando 6 medições por die. O objetivo e identificar **qual fol o maior consumo registrado**. -----------------
### A) O algoritmo está incorreto, pois a variável maior deveria ser inicializada com zero, e não com um valor do vetor.
### B) A comparação feita no algoritmo permite encontrar o menor valor, e não o maior consumo.
### C) A estrutura de repetição Para (linha 13) está indevidamente usada, pois deveria iniciar no índice 1 e não no 2.
### D) O algoritmo lê corretamente os valores de consumo e identifica o maior deles ao final da execução.
### E) O vetor consumo deveria ser do tipo inteiro, já que a variável maior não armazena números com casas decimais.

### **Resposta correta: (D)**
