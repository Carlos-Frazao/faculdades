### 10. Um estacionamento com 10 vagas deseja automatizar o controle de entrada e saída de veículos. A administração pretende usar um vetor de 10 posições para armazenar as placas dos carros estacionados. Você foi contratado para desenvolver um **algoritmo em pseudocódigo** que execute as seguintes operações:

+ ### Permita ao usuário **informar as placas dos veículos** que estão ocupando as 10 vagas (uma placa por posição).

+ ### Após o preenchimento, o sistema deve **ler uma placa** digitada pelo usuário e **verificar se ela está entre as vagas ocupadas**.

+ ### Caso a placa seja encontrada, o sistema deve **informar em qual vaga o carro está** (posição no vetor).

+ ### Caso contrário, deve exibir a mensagem: "Veículo não encontrado no estacionamento.
