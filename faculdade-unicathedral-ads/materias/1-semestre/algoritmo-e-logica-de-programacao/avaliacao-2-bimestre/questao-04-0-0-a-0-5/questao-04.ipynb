Durante o desenvolvimento de um sistema de apoio à coleta seletiva, um estudante criou um algoritmo para somar os valores pares representando o número de itens reciclados em diferentes setores de uma escola. Considere o seguinte trecho de pseudocódigo: Com base no pseudocódigo acima e no contexto apresentado, assinale a alternativa correta:

---------------
### A) O algoritmo calcula a soma de todos os valores ímpares armazenados no vetor.
### B) O comando mod 2 = 0 verifica se o número lido é múltiplo de dois, ou seja, se é par.
### C) O valor da variável soma será sempre 0, já que vetores não podem ser somados dentro de laços.
### D) O algoritmo está incorreto por não inicializar o vetor antes da leitura.
### E) A estrutura condicional Se deve vir antes da leitura para garantir a verificação.

### **Resposta correta: (B)**
