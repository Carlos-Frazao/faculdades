### 02. Em diversos contextos computacionais, como controle de sensores, listas de espera e registros diários, o uso de vetores permite organizar informações de forma linear e eficiente. Assim, o entendimento das características conceituais dos vetores é essencial para a construção de algoritmos corretos. -----------------
### A. Vetores são estruturas de dados que permitem armazenar elementos de diferentes tipos em posições consecutivas de memória.
### B. A principal limitação dos vetores é que os dados armazenados não podem ser acessados diretamente, sendo necessário percorrê-los por completo sempre que se deseja consültar um valor.
### C. Um vetor de 10 posições pode ser utilizado para armazenar informações sobre 10 elementos diferentes, desde que todos sejam do mesmo tipo.
### D. Em pseudocódigo, vetores são sempre dinâmicos. podendo ter seu tamanho alterado durante a execução do programa sem necessidade de redimensionamento.
### E. Para declarar um vetor, é obrigatório utilizar laços de repetição ou funções específicas, mesmo que o vetor seja fixo e seus valores conhecidos.
