### 07. No estudo de algoritmos computacionais, uma das primeiras habilidades desenvolvidas é a capacidade de criar soluções para problemas utilizando **entrada de dados, processamento (cálculos)** e **saída de resultados**. Essa estrutura básica é a essência de algoritmos simples, mas fundamentais para 0 desenvolvimento de sistemas computacionais.

### Um exemplo prático desse conceito está no desenvolvimento de sistemas para pequenas empresas, como um serviço de entrega de gás. O proprietário deseja automatizar o cálculo do valor total cobrado ao cliente, com base em informações simples da venda.

### Crie um **algoritmo computacional** que receba como entrada os seguintes dados:
+ ### Nome do cliente

+ ### Quantidade de botijões de gás adquiridos
### O algoritmo deve **processar** esses dados da seguinte forma:

+ ### Calcular o valor total da compra como:
        **Valor total = (quantidade x 80) + 5**
### Por fim, o algoritmo deve **exibir** o nome do cliente e o valor total a ser pago.
