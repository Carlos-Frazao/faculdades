### 04. Durante uma aula de lógica de programação, o professor propõe uma atividade curiosa: cada grupo de alunos deve escrever um **algoritmo** que represente uma **receita de bolo**. A ideia é que a sequência de passos seja clara o suficiente para que um robô consiga executar a tarefa, do início ao fim, sem ambiguidades. Um dos grupos descreveu o seguinte:

### 1. Pegar os ingredientes (farinha, ovos, açúcar, fermento, leite e manteiga);
### 2. Misturar todos os ingredientes em uma tigela;
### 3. Untar a forma com manteiga e farinha;
### 4. Despejar a massa na forma;
### 5. Colocar no forno pré-aquecido a 180°C por 40 minutos;
### 6. Retirar do forno e deixar esfriar.

### Após a apresentação, o professor pergunta: "De que forma essa receita se relaciona com a construção de algoritmos na programação de computadores?"

### Com base na analogia da receita de bolo e nos fundamentos da lógica de programação, assinale a alternativa que melhor representa a relação entre a receita e um algoritmo computacional:
---------------
### A. A receita representa o código executável de um sistema, onde os ingredientes são os dispositivos de entrada e o forno é a memória RAM.
### B. A receita é a representação de um algoritmo, pois apresenta uma sequência lógica, ordenada e finita de ações para atingir um resultado específico.
### C. A receita representa apenas a etapa de entrada de um algoritmo, pois os ingredientes são coletados antes de qualquer processamento.
### D. A receita é equivalente a linguagem de programação. pois contém todas as informações que serão utilizadas durante a execução de um programa.
### E. A receita representa o compilador, pois transforma instruções humanas em comandos executáveis por uma máquina.

### **Resposta correta: (B)**
