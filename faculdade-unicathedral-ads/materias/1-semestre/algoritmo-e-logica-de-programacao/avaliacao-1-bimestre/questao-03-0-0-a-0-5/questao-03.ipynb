### 03. Uma empresa de transporte urbano decidiu implementar um sistema digital para calcular automaticamente o valor da tarifa que será cobrada de cada passageiro. O cálculo deve considerar a distância percorrida e o tipo de passageiro (comum, estudante ou idoso). O sistema recebe os dados inseridos pelo usuário (tipo de passageiro e distância), aplica a lógica de cálculo de tarifa (considerando descontos e tarifas mínimas) e apresenta o valor final a ser pago.
### Um estagiário, ao observar o funcionamento do sistema, questiona qual seria a estrutura lógica por trás de seu funcionamento. O analista responsável, então, explica que todo algoritmo segue uma estrutura fundamental composta por três etapas: **entrada, processamento e saída**.
### Considerando o cenário descrito e os princípios da lógica de programação, assinale a alternativa que melhor representa as três etapas fundamentais da estrutura de um algoritmo:
-----------------
### A. **Entrada**: exibir o valor da tarifa; **Processamento**: calcular o tipo de usuário; **Saída**: receber os dados do passageiro.
### B. **Entrada**: calcular a distância; **Processamento**: aplicar o desconto; **Saída**: armazenar os dados no banco.
### C. **Entrada**: receber os dados do passageiro; **Processamento**: aplicar regras de cálculo sobre os dados recebidos; **Saída**: apresentar o valor final da tarifa.
### D. **Entrada**: definir as variáveis internas: **Processamento**: mostrar o resultado; **Saida**: executar a lógica do programa.
### E. **Entrada**: importar bibliotecas do sistema; **Processamento**: interpretar o código; **Saída**: encerrar a execução do algoritmo.

### **Resposta correta: (C)**
