### 05. Durante uma entrevista para uma vaga de estágio em desenvolvimento de software, o candidato foi desafiado com a seguinte pergunta pelo avaliador técnico:

+ ### "Se você tivesse que explicar o que é um **algoritmo** para alguém que nunca programou na vida, como faria isso?"

### O candidato respondeu: 
+ ### "Eu diria que um algoritmo é como uma **receita de bolo**: ela define passo a passo o que fazer, em qual ordem, e com quais ingredientes, para que o resultado final seja sempre o mesmo, independente de quem esteja executando."
### O avaliador, satisfeito com a resposta, pediu então para o candidato identificar quais aspectos da receita podem ser relacionados com os **elementos essenciais de um algoritmo**.
### Considerando a analogia entre uma **receita culinária** e um **algoritmo computacional**, correspondências a seguir: analise as correspondências a seguir:

### I. Ingredientes representam os **dados de entrada**.
### II. O modo de preparo representa a **sequência lógica de processamento**.
### III. O bolo pronto representa a **saída (resultado final)**.
### Assinale a alternativa correta:
--------
### A. Apenas a afirmativa I está correta.
### B. Apenas a afirmativa Il está correta.
### C. Apenas a afirmativa III está correta.
### D. Todas as afirmativas estão corretas.
### E. Nenhuma das afirmativas está correta.

### **Resposta correta: (D)**
