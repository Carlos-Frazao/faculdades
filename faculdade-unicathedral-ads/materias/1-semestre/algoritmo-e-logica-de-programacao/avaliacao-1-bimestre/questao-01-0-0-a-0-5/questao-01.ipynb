### 01. Os algoritmos são a espinha dorsal da tecnologia moderna, trabalhando nos bastidores para tornar nossa vida mais fácil, mais segura e mais agradável, oferecendo soluções personalizadas e otimizando processos em vários setores. Suas aplicações são praticamente ilimitadas.
### No contexto da programação de computadores, um algoritmo pode ser definido como:
--------
### A. Um equipamento responsável por executar tarefas automáticas a partir de instruções digitais.
### B. Um conjunto de códigos binários que controlam os dispositivos de hardware.
### C. Uma sequência de instruções lógicas, ordenadas e finitas que visam resolver um problema ou realizar uma tarefa.
### D. Um programa executável que interpreta linguagens de alto nível.
### E. Uma linguagem de programação utilizada para automatizar processos computacionais
### **Respostas corretas: (C)**
