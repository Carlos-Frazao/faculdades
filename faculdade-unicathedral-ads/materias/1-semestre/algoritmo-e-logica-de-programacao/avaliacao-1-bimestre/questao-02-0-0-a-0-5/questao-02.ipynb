### 02. Uma equipe de desenvolvedores está criando um sistema para uma academia, que precisa calcular o indice de massa corporal de seus alunos. Para isso, o programa deve receber dados como o **peso** e a **altura** de cada pessoa, realizar o cálculo e, em seguida, exibir o resultado junto à classificação (ex: abaixo do peso peso ideal, sobrepeso etc.).
### Durante o desenvolvimento, um dos membros da equipe pergunta: "Como o computador sabe onde armazenar e recuperar os valores de peso e altura durante o processamento?" O coordenador do projeto, então, começa a explicar o papel das **variáveis** no algoritmo.
### No contexto da construção de algoritmos computacionais, as **variáveis** desempenham um papel fundamental. Com base no exemplo apresentado, é correto afirmar que uma variável:
-----------------
### A. É uma constante utilizada para armazenar dados fixos que não podem ser modificados durante a execução do algoritmo.
### B. Representa uma função matemática que substitui a necessidade de cálculos manuais no algoritmo.
### C. É um espaço de memória reservado para armazenar temporariamente dados que podem ser manipulados ao longo da execução do algoritmo.
### D. É o nome dado ao algoritmo como um todo, responsável por identificar sua função principal.
### E. Refere-se à interface gráfica responsável por exibir as informações armazenadas em tempo real.

### **Resposta correta: (C)**
