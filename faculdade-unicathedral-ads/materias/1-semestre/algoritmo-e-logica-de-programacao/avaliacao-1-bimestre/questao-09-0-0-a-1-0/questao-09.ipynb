### 09. Ao aprender algoritmos computacionais, um dos conceitos fundamentais é a **estrutura de repetição**, utilizada quando uma ação precisa ser executada múltiplas vezes. Esse recurso permite processar uma quantidade variável de dados sem necessidade de reescrever instruções.
### Imagine um cenário em que uma papelaria deseja automatizar o registro de vendas realizadas durante o dia. Para isso, será necessário criar um algoritmo que repita o processo de cálculo de vendas para cada cliente atendido, até que um comando de parada seja inserido.
### Crie um algoritmo computacional que funcione da seguinte forma:
### 1. Para cada cliente atendido, o sistema deve:          
+ + ### Solicitar o nome do cliente
+ + ### Solicitar o valor da compra
### 2. Calcular e armazenar o valor total acumulado de todas as compras
### 3. O sistema deve continuar atendendo clientes até que seja digitado o nome "FIM" no campo do nome.
### 4. Ao final, exibir:
+ + Ο O número total de clientes atendidos
+ + O valor total acumulado do dia.
