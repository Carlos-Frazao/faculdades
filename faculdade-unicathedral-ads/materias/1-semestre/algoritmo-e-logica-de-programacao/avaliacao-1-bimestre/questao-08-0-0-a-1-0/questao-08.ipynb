### 08. Em uma escola de idiomas, os alunos realizam uma **prova com três** seções: leitura, escrita e conversação. Cada seção vale de **0 a 10 pontos**. Para ser aprovado no curso, o aluno precisa obter **nota igual ou superior a 7 em cada uma das três seções**.
### A coordenação do curso solicitou a criação de um pequeno sistema que **avalie automaticamente** a situação de cada aluno com base nas notas inseridas. Com base na situação apresentada, elabore um algoritmo que:
### 1. Solicite ao usuário que insira as notas de **leitura, escrita e conversação**; ###
### 2. Verifique se o aluno foi **aprovado** ou **reprovado** (lembrando que é necessário ter **nota ≥ 7** em todas as seções);
### 3. Mostre a média das três notas e a situação final do aluno: "**Aprovado**" ou "**Reprovado**".
### Escreva seu algoritmo utilize estruturas básicas como variáveis, entrada/saída de dados, e estruturas condicionais.
