### 01. Em Python, funções são blocos de código reutilizáveis que podem receber dados de entrada e, opcionalmente, retornar um resultado. Elas desempenham um papel fundamental na organização de programas, facilitando a modularização, a legibilidade e a manutenção do código. Com base nesse conceito, analise as afirmações a seguir:

+ ### I. Toda função em Python é obrigada a retornar um valor.X

+ ### II. O uso de funções permite que um mesmo trecho de código seja executado várias vezes, sem a necessidade de duplicação.

+ ### III. Funções podem receber dados por meio de parâmetros, que são declarados entre parênteses durante sua definição.

+ ### IV. Uma função definida em Python não será executada até que seja chamada em algum ponto do programa.

### Assinale a alternativa correta:

### A) Apenas II e III estão corretas.

### B) Apenas II, III e IV estão corretas.

### C) Apenas I e IV estão corretas.

### D) Todas estão corretas.

### E) Nenhuma está correta.

### **Respostas corretas: (C)**
