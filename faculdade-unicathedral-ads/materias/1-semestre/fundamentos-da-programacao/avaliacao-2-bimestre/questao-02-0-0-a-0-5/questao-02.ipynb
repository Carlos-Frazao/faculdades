### 02. Em um projeto interdisciplinar, estudantes de um curso de Engenharia desenvolveram um programa para auxiliar colegas na conversão de unidades de distância. A ideia era que o usuário informasse uma medida em quilômetros, e o programa retornasse o valor equivalente em milhas, utilizando uma função com parâmetro e retorno.


```python
def converter_km_para_milhas(km):
    return km * 0.621371

distancia = float(input("Informe a distância em Km: "))
resultado = converter_km_para_milhas(distancia)

print(f"A distância em milhas é: {resultado:.2f}")
```

+ ### III. O valor retornado pela função é armazenado na variável resultado, que pode ser usado em outras partes do programa.C

+ ### IV. Se a função utilizasse print() no lugar de return, ela continuaria retornando o valor convertido. X
