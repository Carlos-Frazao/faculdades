### 04. Um programa armazena em uma lista os dias da semana. e o usuário informa um número de 1 a 7 para saber qual é o dia correspondente. No entanto, o trecho de código que faz essa associação ainda está incompleto. Veja o início do código:


```python
def main():
    dias = ["domingo", "segunda", "terça", "quarta", "quinta", "sexta", "sábado"]
    numero = int(input("Digite um número de 1 a 7: "))

    # Trecho que falta aqui
    if 1 <= numero <= 7:
        print(dias[numero - 1])
    else:
        print("Número inválido! Por favor, digite um número entre 1 e 7.")
    # Fim do trecho que falta

main()
```
