### 03. Durante a elaboração de um sistema de controle de tomperatura para laboratórios universitários, estudante do curso de Ciência da Computação implementou a seguinte função para converter a temperatura de Celsius para Fahrenheit:


```python
def converter(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit
```

### Com base nessa função, outro estudante escreveu o seguinte trecho de código:

```python
def exibir_temperatura(temp): 
    print("Temperatura converter(temp)) atual em Fahrenheit:",converter(temp))

temperatura_ambiente = float(input("Informe a temperatura em Celsius: "))

exibir_temperatura(temperatura_ambiente)
```
