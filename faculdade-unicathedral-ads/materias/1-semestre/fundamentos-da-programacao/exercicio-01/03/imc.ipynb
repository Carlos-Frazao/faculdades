### 03. Construa um algoritmo para determinar se o indivíduo está com um peso favorável. Essa situação é determinada através do IMC (ÍNDICE de MASSA CORPÉREA), que é definida como sendo a relação entre o peso (PESO) e o quadrado da altura (ALTURA) do indivíduo. Ou seja, <br>
$$IMC = \frac{PESO}{ALTURA^{2}}$$
### e, a situação do peso é determinado pela tabela abaixo.
$$
\begin{array}{|c|c|}
\hline 
Condição                    &   Situação            \\
\hline
IMC \ abaixo \ de \ 20      &   Abaixo \ do \ Peso  \\
\hline
IMC \ de \ 20 \ até \ 25    &   Peso \ Normal       \\
\hline
IMC \ de \ 25 \ até \ 30    &   Sobre \ Peso        \\
\hline
IMC \ de \ 30 \ té \ 40     &   Obeso               \\
\hline
IMC \ de \ 40 \ e \ acima   &   Obeso Móbido        \\
\hline
\end{array}
$$

```python
peso = float(input("Digite seu Peso:")) # Pedindo o peso.
altura = float(input("Digite sua Altura: ")) # Pedindo a altura.
imc = peso / (altura * altura)
if imc < 20: # Analisando as condições e situações.
    print(f"Seu IMC é {imc}, sua situação está Abaixo do peso!")
elif imc >= 20 and imc < 25:
    print(f"Seu IMC é {imc}, Peso normal.")
elif imc >= 25 and imc < 30:
    print(f"Seu IMC é {imc}, Sobre peso!")
elif imc >= 30 and imc < 40:
    print(f"Seu IMC é {imc}, Obeso!")
else:
    print("Obeso Móbido!")
```
