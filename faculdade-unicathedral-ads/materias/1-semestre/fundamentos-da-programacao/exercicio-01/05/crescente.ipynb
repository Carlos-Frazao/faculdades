### 05. Construa um algoritmo que imprima em ordem crescente dois números A a B, lidos através do teclado.

```python
a = int(input("Digite um número: ")) # Pedindo número.
b = int(input("Digite outro número: ")) # Pedindo número.
if a > b: # Verificando se o A é maior que o B, se for deverá retornar(B, A) 
    print(f"{b}:{a}")
else:
    print(f"{b}:{a}")
```
