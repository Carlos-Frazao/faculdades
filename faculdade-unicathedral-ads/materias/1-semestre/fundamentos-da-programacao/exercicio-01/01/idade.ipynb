### 01. Escreva um algoritmo que receba a idade de uma pessoa e então apresente uma mensagem dizendo se ela é maior ou menor de idade.

```python
idade = int(input("Digite sua idade: ")) # Pedindo ao usuário inserir sua idade.
if idade >= 18: # Analisando se a idade inserida é maior ou igual a 18.
    print("Você é maior de idade.")
else:
    print("Você é menor de idade.")
```
