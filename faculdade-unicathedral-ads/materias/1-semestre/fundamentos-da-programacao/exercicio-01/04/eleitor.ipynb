### 04. No sistema eleitoral brasileiro temos 3 categorias de eleitores:
+ Não votantes: pessoas abaixo de 16 anos;
+ Votantes obrigatórios: pessoas de 18 até 70.
+ Votantes opcionais: pessoas de 16 anos e abaixo de 18 anos e para pessoas de 70 anos.
### Escreva um algoritmo que receba o valor da idade e então apresente a situação de voto dessa pessoa.

```python
idade = int(input("Digite sua idade: ")) # Pedindo a idade para analizar em que categoria esse eleitor estará.
if idade < 16: # Analisando a condição de pessoas não votantes.
    print("Não votantes.")
elif idade >= 18 and idade < 70:  # Analisando a condição de pessoas Obrogadas a votar.
    print("Votantes obrigatórios.")
elif idade >= 16 and idade <= 18 and idade >= 70:  # Analisando a condição das pessoas que são votantes opcionais de 16 á 18 e > 70.
    print("Votantes opcionais")
```
