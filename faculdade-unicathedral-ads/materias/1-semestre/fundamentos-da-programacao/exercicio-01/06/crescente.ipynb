### 06. Construa um algoritmo que imprima em ordem crescente três números A, B e C, lidos através do teclado.

```python
a = int(input("Digite um número: ")) # Pedindo número.
b = int(input("Digite outro número: ")) # Pedindo número.
c = int(input("Digite outro número: ")) # Pedindo número.

if a > b and a > c: # Verificando se o A é maior que o B e o C.
    if b > c: # Verificando se o B é maior que o C.
        print(f"{c}:{b}:{a}") 
    elif c > b: # Verificando se o C é maior que o B.
        print(f"{b}:{c}:{a}")

elif b > a and b > c:
    if a > c:
        print(f"{c}:{a}:{b}")
    elif c > a:
        print(f"{a}:{c}:{b}")

elif c > a and c > b:
    if a > b:
        print(f"{b}:{a}:{c}")
    elif b > a:
        print(f"{a}:{b}:{c}")

else:
    pass
    #print(f"{b}:{a}")
```
