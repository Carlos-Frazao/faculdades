### 04. O laço de repetição while é utilizado para repetir um bloco de código enquanto uma condição for verdadeira. É muito útil quando o número de repetições não é previamente conhecido ou depende de uma condição variável.

### Considere o seguinte código em Python:

```python
contador = 1
total = 0

while contador <= 4:
    total = total + contador
    contador = contador + 1

print("Total:", total)
```

### Após a execução do código, qual será a saída exibida no console?
------------
### A. Total: 10
### B. Total: 6
### C. Total: 5
### D. Total: 4
### E. O programa entra em loop infinito
### **Resposta correta: (A)**
