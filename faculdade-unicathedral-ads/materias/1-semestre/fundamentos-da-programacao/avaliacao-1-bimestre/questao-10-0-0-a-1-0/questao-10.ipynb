### 10. A estrutura de repetição while em Python permite executar blocos de código repetidamente, desde que uma condição seja verdadeira. É comumente usada quando o número de repetições não é conhecido previamente, como no caso de validações de entrada ou somatórios progressivos.
### Desenvolva um algoritmo que some números inteiros positivos fornecidos pelo usuário. O programa deve continuar lendo números até que o usuário digite 0, momento em que a soma total deve ser exibida.

```python
#Maneira correta.
n = 1
s = 0

while n != 0:
    n = int(input("Informe um número inteiro (0 para sair): "))
    s += n
    if n < 0:
        print("Número inválido, digite um número positivo.")
        s -= n #Essa linha tbm pode ser escrita como s = s - n.
print(f"A soma total dos números é: {s}") 
```
