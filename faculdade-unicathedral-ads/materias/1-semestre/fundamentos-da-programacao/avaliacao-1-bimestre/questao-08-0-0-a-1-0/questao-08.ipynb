### 08. Verificações binárias são comuns em sistemas informatizados que precisam tomar decisões a partir de um único critério lógico. A estrutura if-else é ideal para lidar com esse tipo de situação, permitindo que o programa execute um bloco de código ou outro, com base no resultado da condição avaliada.
### Um sistema de cadastro de motoristas precisa identificar se um cidadão tem idade suficiente para obter a Carteira Nacional de Habilitação (CNH). Esse tipo de verificação é um exemplo clássico do uso de condicionais simples no desenvolvimento de sistemas administrativos.
### Implemente um programa em Python que:

+   Solicite a **idade** do usuário;
+   Caso tenha **18 anos ou mais**, exiba: "Você pode solicitar a CNH";
+   Caso contrário, exiba: "Você ainda não tem idade para solicitar a CNH".

```python
idade = int(input("Informe a sua idade: "))
if idade >= 18:
    print("Você pode tirar a CNH.")
else:
    print("Você não pode tirar a CNH.")
```
