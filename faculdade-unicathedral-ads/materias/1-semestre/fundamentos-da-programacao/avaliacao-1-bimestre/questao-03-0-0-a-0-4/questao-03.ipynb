### 03. Analisar a saída de um trecho de código é uma habilidade essencial para identificar o funcionamento de algoritmos, especialmente em situações de tomada de decisão. Em Python, a estrutura if-else permite executar diferentes blocos de código conforme uma condição seja verdadeira ou falsa.

### Considere o seguinte trecho de código Python:

```python
x = 10
y = 5

if x > y:
    resultado = x - y

else:
    resultado = y - x

print("Resultado:", resultado)
```

### Com base no código acima, qual será a saída exibida no console?
--------------
### A. Resultado: 5
### B. Resultado: -5
### C. Resultado: 15
### D. Resultado: 0
### E. O código apresentará erro de execução
### **Resposta correta: (A)**
