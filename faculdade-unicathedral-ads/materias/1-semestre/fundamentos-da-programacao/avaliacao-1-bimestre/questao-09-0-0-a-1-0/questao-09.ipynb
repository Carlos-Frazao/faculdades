### 09. No desenvolvimento de aplicações com Python, as estruturas condicionais (if, elif, else) são essenciais para a tomada de decisões automáticas baseadas em valores de entrada. Esses comandos utilizam operadores relacionais (como >, <, ==) para avaliar condições lógicas que determinam o fluxo do programa.
### Crie um programa em Python que leia a idade de uma pessoa e classifique-a conforme os seguintes critérios:

+   **Menor**: menos de 18 anos
+   **Adulto**: entre 18 e 59 anos
+   **Idoso**: 60 anos ou mais

### O programa deve exibir a categoria correspondente com uma mensagem apropriada.

```python
idade = int(input("Digite a sua idade: "))
if idade < 18:
    print("Você é menor de idade.")
elif idade >= 18 and idade <= 59:
    print("Você é adulto.")
else:
    print("Você é idoso.")
```
