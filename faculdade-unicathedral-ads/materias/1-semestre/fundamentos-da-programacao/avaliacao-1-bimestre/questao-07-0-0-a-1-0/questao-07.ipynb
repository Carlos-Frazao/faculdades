### 07.Em Python, funções como input() são utilizadas para entrada de dados, mas seu retorno é sempre do tipo str (texto). Para realizar cálculos, é necessário converter essas entradas para tipos numéricos como inteiros ou reais, usando funções como int() e float().
### Implemente um programa em Python que leia dois números reais informados pelo usuário, calcule a soma deles e exiba o resultado formatado com duas casas decimais.

```python
n1 = float(input("Informe um número, real inteiro: "))
n2 = float(input("Informe outro número, real inteiro: "))
soma = n1 + n2

print(f"A soma é: {soma:.2f}")
```
