### 02. Uma das principais aplicações das estruturas condicionais em Python está na **validação de regras de negócio**, como em sistemas bancários, educacionais ou comerciais. Quando há múltiplas condições a serem avaliadas, o uso de operadores lógicos como and e or se torna fundamental para garantir decisões corretas.

### Um desenvolvedor está criando um sistema para verificar se um cliente tem direito a um **empréstimo bancário**. O banco concede empréstimo apenas para pessoas com **idade entre 18 e 65 anos e renda mensal acima de R$ 2.000,00**.
### Considere o seguinte código:


```python
idade = int(input("Digite sua idade: "))
renda = float(input("Digite sua renda mensal: "))

if idade >= 18 and idade <= 65 and renda > 2000: 
    print("Empréstimo aprovado.")
else:
    print("Empréstimo negado.")
```

### Com base nesse cenário, assinale a alternativa correta:
-------
### A. O código está incorreto porque a estrutura condicional só aceita uma única condição por vez.
### B. A mensagem "Empréstimo negado" será exibida sempre que a idade for menor que 18 ou a renda for inferior a 2000.
### C. A função float() não pode ser usada junto com input() para valores monetários. K
### D. O operador and faz com que todas as condições sejam ignoradas, pois apenas a última é avaliada.
### E. O código está incorreto porque o operador <= não pode ser usado com variáveis
### **Resposta Correta: (B)**
