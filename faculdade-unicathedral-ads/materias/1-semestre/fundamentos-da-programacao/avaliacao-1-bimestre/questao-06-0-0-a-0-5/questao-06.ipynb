### 06. No desenvolvimento de linguagens de programação, um dos aspectos fundamentais é a forma como a estrutura dos blocos de código é definida. Linguagens como C, Java e JavaScript utilizam símbolos como chaves {} para delimitar blocos, permitindo certa flexibilidade na disposição visual do código. Já a linguagem Python segue uma filosofia diferente: ela valoriza fortemente a ***clareza e a legibilidade** do código.

### No contexto da linguagem Python, a **indentação** exerce um papel fundamental na organização dos blocos de código. Sobre a indentação em Python, assinale a alternativa correta:
--------------
### A. indentação em Python é apenas uma recomendação de estilo, sendo opcional para a execução do programa.
### B. Python permite que o programador escolha entre usar indentação ou ponto e vírgula para organizar blocos de código.
### C. A ausência de indentação pode ser compensada pelo uso de delimitadores como chaves { } ou colchetes [].
### D. A indentação correta é obrigatória para que o Python compreenda onde um bloco de código começa e termina.
### E. A indentação só é exigida quando o código está dentro de estruturas de repetição, sendo dispensável em condicionais.
### **Resposta correta: (D)**
