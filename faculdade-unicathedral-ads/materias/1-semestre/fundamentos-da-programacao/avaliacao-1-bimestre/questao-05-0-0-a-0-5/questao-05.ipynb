No entanto, isso também exige que o programador seja rigoroso com os recuos, principalmente em **estruturas aninhadas** como if dentro de if, ou if-else combinados com outros comandos.

### Considere o seguinte código Python:

```python
numero = int (input("Digite um número: "))

if numero > 0:
print("Positivo")

if numero % 2 == 0:
    print("Par")
else:
    print("Ímpar")

elif numero < 0:
    print("Negativo")
else:
    print("Zero")
```
