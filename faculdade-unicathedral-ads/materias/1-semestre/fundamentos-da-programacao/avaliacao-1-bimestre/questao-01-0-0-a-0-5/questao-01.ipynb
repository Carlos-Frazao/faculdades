### 01. Um dos primeiros passos no desenvolvimento de programas com Python é compreender como o usuário pode fornecer dados ao sistema e como o programa pode interpretá-los e exibi-los de forma adequada. Para isso, é essencial conhecer os tipos primitivos (como int, float, str e bool) e saber realizar a conversão correta desses valores.

### Considere o seguinte trecho de código em Python:


```python
nome = input("Digite seu nome: ")

idade = input("Digite sua idade: ")

altura = input("Digite sua altura: ")

print(f"{nome} tem {idade} anos e mede {altura} metros.")
```

### Esse código funciona, mas pode apresentar problemas caso o programador precise realizar operações matemáticas com os dados fornecidos. Assinale a alternativa correta em relação à manipulação de tipos primitivos nesse contexto.
-----------------

### A. O código está correto, pois 0 Python automaticamente detecta e converte os tipos de dados inseridos pelo usuário.
### B. As variáveis idade e altura devem ser convertidas para str para permitir a formatação com f-strings.
### C. A função input() em Python sempre retorna um valor do tipo float,, por isso o código funcionará corretamente.
### D. Para realizar cálculos com os valores inseridos, é necessário converter explicitamente idade para int e altura para float.
### E. O comando print() não aceita variáveis de tipos diferentes em uma mesma f-string, sendo necessário convertê-las para string manualmente
### **Resposta correta: (D)**
