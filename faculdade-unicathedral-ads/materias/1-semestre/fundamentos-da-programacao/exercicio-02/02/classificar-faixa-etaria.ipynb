### 02. Crie uma função chamada classificar_faixa_etaria(idade) que receba a idade de uma pessoa (inteiro) e retorne uma string com a classificação:

+ ### 0 a 12 anos: "Infantil"

+ ### 13 a 17 anos: "Adolescente"

+ ### 18 a 59 anos: "Adulto"

+ ### 60 anos ou mais: "Idoso"


```python
def ClassificarFaixaEtaria(i): #i vale idade.
    if i <= 12:
        return "Infantil"
    elif i >= 13 and i <= 17:
        return "Adolecente"
    elif i >= 18 and i <= 59:
        return "Adulto"
    else:
        return "Idoso"
    
idade = int(input("Informe sua idade: "))

print(ClassificarFaixaEtaria(idade))
```
