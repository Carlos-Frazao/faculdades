### 05. Você está criando um jogo simples de adivinhação onde o computador escolhe um número aleatório entre 1 e 100, e o jogador deve adivinhar qual é esse número. O jogo deve fornecer dicas, informando se o número escolhido é maior ou menor do que o palpite do jogador, até que o jogador acerte.

### Crie uma função chamada adivinhar() que faça o seguinte:

+ ### 1. Gere um número aleatório entre 1 e 100.

+ ### 2. Solicite ao jogador que faça um palpite.

+ ### 3. Diga se o palpite é maior, menor ou igual ao número gerado.

+ ### 4. O jogo deve continuar até que o jogador acerte o número.

+ ### 5. Quando o jogador acertar, o jogo deve exibir uma mensagem dizendo que ele ganhou, juntamente com o número de tentativas feitas.

### Use o módulo random para gerar o número aleatório (random.randint(1.100)),

```python
import random
numero_escolhido = random.randint(1, 100) # Gerando o número aleatório.
def funcao_advinha():
    tentativa = 0
    numero_teste = 0
    print("Bem-vindo ao jogo de adivinhação!")
    print("Estou pensando em um número entre 1 e 100.")
    while numero_escolhido != numero_teste:
        palpite_jogador = int(input("Qual número estou pensando? "))
        print(numero_escolhido)
        if palpite_jogador == numero_escolhido:
            print("Você acertou!")
            numero_teste = numero_escolhido # Atribuindo o numero teste para receber o numero escolhido. Para sair do loop.
        elif palpite_jogador < numero_escolhido:
            print("O número que você digitou é menor que o número escolhido pelo computador.")
        else:
            print("O número que você digitou é maior que o número escolhido pelo computador.")
        tentativa += 1
    print(f"Você tentou {tentativa} vez(es).")
funcao_advinha() 
```
