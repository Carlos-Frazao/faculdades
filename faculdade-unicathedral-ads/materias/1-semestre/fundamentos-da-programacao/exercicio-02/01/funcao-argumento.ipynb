### 01. Faça um programa, com uma função que necessite de um argumento. A função retorna o valor de caractere 'P', se seu argumento for positivo, e 'N', se seu argumento for negativo e 'O' se seu argumento for zero.


```python
def verificação(numero):
    if numero > 1:
        return "P"
    elif numero == 0:
        return "O"
    else:
        return "N"

n = float(input("Informe um número: "))

print(verificação(n))
```
