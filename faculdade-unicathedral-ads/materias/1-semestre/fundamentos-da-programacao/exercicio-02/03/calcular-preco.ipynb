### 03. Crie uma função chamada calcular_preco_final(valor, quantidade) que receba:

+ ### valor: valor do produto (float)

+ ### quantidade: número de itens comprados (int)

### A função deve aplicar o seguinte desconto conforme a quantidade de itens:

+ ### 1 a 2 itens: **0%** de desconto

+ ### 3 a 4 itens: **10%** de desconto

+ ### 5 itens ou mais: **20%** de desconto

### A função deve retornar o valor final da compra, já com o desconto aplicado.



```python
def calcular_preco(valor, quantidade):
    if quantidade <= 2:
        return valor * quantidade
    elif quantidade <= 4:
        desconto = valor - (valor * 10 / 100)
        return desconto
    else:
        desconto = valor - (valor * 20 / 100)
        return desconto

produto = float(input("Informe o valor do produto: "))
qtidade = int(input("Informe a quantidade de produtos comprados: "))

print(f"O valor final da compra é :{calcular_preco(produto, qtidade)}")
```
